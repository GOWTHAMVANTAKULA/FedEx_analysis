````python
#outliers of every column
def calculate_outliers(data, columns):
    outliers = {}
    for col in columns:
      data[col] = pd.to_numeric(data[col], errors='coerce')# convert all values to numeric and all missing values to NaN
      Q1 = data[col].quantile(0.25)  # First quartile (25%)
      Q3 = data[col].quantile(0.75)  # Third quartile (75%)
      IQR = Q3 - Q1
      lower_bound = Q1 - 1.5 * IQR
      upper_bound = Q3 + 1.5 * IQR
      lower_bound = max(lower_bound, 0)
      lower_outliers = data[data[col] < lower_bound][col]
      upper_outliers = data[data[col] > upper_bound][col]
      outliers[col] = {
            'lower_bound': lower_bound,
            'upper_bound': upper_bound,
            'lower_outliers': lower_outliers,
            'upper_outliers': upper_outliers
        }
    return outliers

numeric_columns = ['Unit of Measure (Per Pack)', 'Line Item Quantity',
                   'Line Item Value', 'Pack Price', 'Unit Price',
                   'Weight (Kilograms)', 'Freight Cost (USD)',
                   'Line Item Insurance (USD)']

outliers = calculate_outliers(data, numeric_columns)

for col, data_out in outliers.items():
    print(f"Column: {col}")
    print(f"  Lower bound: {(data_out['lower_bound'])}")
    print(f"  Upper bound: {data_out['upper_bound']}")
````

````python
delivery_rates = pd.DataFrame({
    'On-Time': ontime_delivery_rate_by_country,
    'Late':late_delivery_rate_by_country
})
# filter out the countries with 100% ontime delivery
delivery_rates_filtered = delivery_rates[delivery_rates['On-Time']<100].sort_values(by = 'Late',ascending = False)
# Set the width of the bars
bar_width = 0.35
# Set the positions of the bars on the x-axis
bar1 = np.arange(len(delivery_rates_filtered.index))
bar2 = [x + bar_width for x in bar1]
# Create the bar chart
plt.figure(figsize=(12,10))
plt.bar(bar1, delivery_rates_filtered['On-Time'], color='green', width=bar_width, label='On-Time')
plt.bar(bar2, delivery_rates_filtered['Late'], color='red', width=bar_width, label='Late')
# Add labels, title, and legend
plt.xlabel('Country', fontsize=14)
plt.ylabel('Delivery Rate (%)', fontsize=14)
plt.title('On-Time vs. Late Delivery Rates by Country', fontsize=16)
plt.xticks([r + bar_width/2 for r in range(len(delivery_rates_filtered.index))], delivery_rates_filtered.index,rotation=90,ha="center", fontsize=12)  # Center x-axis labels
plt.yticks(fontsize=12)
plt.legend(fontsize=12)
plt.tight_layout() #prevent labels from getting cut off
plt.show()
````

##### 1. Why did you pick the specific chart?

I choosed this grouped bar chart to show the ontime and late deliveries on different countries. This makes it easy to quickly grasp the relative performance of each country in terms of on-time delivery.

##### 2. What is/are the insight(s) found from the chart?
